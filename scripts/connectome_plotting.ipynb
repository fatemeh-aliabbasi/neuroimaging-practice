```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#defining the dk.csv files location
subject = '0012'
dkcsv = f'../derivative/{subject}/dk.csv'
```

```python
# Load connectome matrix
df = pd.read_csv(dkcsv, header=None)
```

```python
# Convert to NumPy array
matrix = df.to_numpy()
```

```python
# Plot heatmap
# Log-transform for better contrast, data-driven colormap range
matrix_log = np.log1p(matrix)   # log(1 + x) handles zeros safely
vmax = np.percentile(matrix_log[matrix_log > 0], 95)

plt.figure(figsize=(10,10))
plt.imshow(matrix_log, vmin=0, vmax=vmax , cmap='viridis')
plt.colorbar(label='log(1 + SIFT2 weight)')
plt.title(f'Structural Connectome — {subject}')
plt.tight_layout()
plt.show()
```
